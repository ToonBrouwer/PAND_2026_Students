### 2.8 — Exploring the Central Limit Theorem

The goal of this problem is to explore the **central limit theorem (CLT)**. We informally
defined the CLT in this chapter. Here, you will perform a numerical experiment to test it.

**a.** Generate a large set of independent random numbers.  
There are many choices, but for example you can generate 500 uniformly distributed
random numbers using MATLAB (e.g., `R = rand(500,1)`).  
Describe this list of random numbers and include a **labeled histogram**.

**b.** Compute the **mean** of this list of numbers.  
What value do you obtain? Does this mean make sense when compared to the histogram
you created?

**c.** Generate another instance of the random numbers from part (a) and compute its mean.  
Repeat this **5,000 times**, saving the mean each time.  
You should end up with **5,000 mean values**, each corresponding to one realization of the
variable \(R\).

Describe the **distribution** of these 5,000 mean values.  
What do you observe?  
Is the result consistent with the CLT? (Hint: create a histogram and examine its shape.)

**d.** Repeat the analysis using a **different distribution** of independent random numbers.  
MATLAB provides many options.

For your new choice of random numbers:

- Describe a **single instance of 500 values** (as in part a)  
- Then describe the **distribution of the mean** of these random numbers (as in part c)

Compare your findings to the results from the uniform distribution.


### 2.9 (Very very Optional, only do if you have time): Cluster based permutation testing
Timeseries form a challenge for statistical testing. Why? 
Well if we have 1000 timepoints and we perform a statistical test at every timepoint we are quickly running into multiple comparison issues (more on this in an optional excercise in the practical on spiking analysis). One way to solve this is by performing a multiple comparison correction such as the bonferonni method, however this is overly stringent and can lead to a lot of false negatives. To solve this issue a new statistical procedure was developed: the cluster based permutation test. 
You can read about this procedure in the following paper: https://www.sciencedirect.com/science/article/pii/S0165027007001707
The challenge here is to implement this procedure to test when two EEG files significanlty differ from each other(use the same files as you used in question 2.7).
